**Background**

Extratropical cyclones (ETCs) are large-scale low-pressure systems that develop in the mid-latitude regions. These systems can travel thousands of kilometres and can last several days and are often, but not always, associated with high winds and heavy rain [[Walker+20](https://doi.org/10.1002/wea.3861); [Ulbrich+09](https://doi.org/10.1007/s00704-008-0083-8); [Catto+18](https://doi.org/10.1175/JCLI-D-17-0746.1)]. The most powerful ETCs can cause significant socioeconomic damage [[Hawcroft+12](https://doi.org/10.1029/2012GL053866); [Garnier+18](https://doi.org/10.1016/j.coastaleng.2017.06.014)], with large storm surges associated with ETCs causing loss of life through coastal flooding, and strong winds causing falling of trees and debris, in addition to disrupting transport systems and severely damaging property. They are also a major loss to the insurance sector. To help the sector better understand this risk, this dataset, the Enhanced Windstorm Service (EWS) has been developed. **this is largely from Walker et al 2020**.

ETCs occury in very diverse synoptic situations with some being confined to lower-tropospheric levels and others extending through great depth, with this complexity revealing why there is no single commonly agreed upon scientific definition of what an ETC is and why there are numerous diverse tracking methods; each tracking algorithm has a set of known obstacles to overcome when trying to identify ETCs within the model and observational data [[Neu+13](https://doi.org/10.1175/BAMS-D-11-00154.1)]. This dataset uses two tracking algorithms TempestExtremes [[TE, Ullrich+21](https://doi.org/10.5194/gmd-14-5023-2021)], complementing the one already used in the original service (TRACK, [[Hodges+95](https://doi.org/10.1175/1520-0493%281995%29123%3C3458%3AFTOTUS%3E2.0.CO%3B2); [Hodges+99](https://doi.org/10.1175/1520-0493(1999)127%3C1362:ACFFT%3E2.0.CO;2); [Hoskins and Hodges+02](https://doi.org/10.1175/1520-0469(2002)059%3C1041:NPOTNH%3E2.0.CO;2)]).

THe TempestExtremes algorithm is run considering ERA5 reanalysis [[Hersbach+20](https://doi.org/10.1002/qj.3803)] 6-hourly mean sea level pressure. On the other hand, the Hodges tracking algorithm is run considering a 3-hourly frequency 850 hPa relative vorticity dataset, also from the ERA5 reanalysis. The input is on the native IFS gaussian grid (N320). Using these two different approaches in identification can lead to variations in the outputted storm track statistics. One reason for this is that results using MSLP represent the low-frequency, large-scale features of the atmosphere, whereas vorticity represents the high-frequency, small-scale features [[Hoskins and Hodges+02](https://doi.org/10.1175/1520-0469(2002)059%3C1041:NPOTNH%3E2.0.CO;2); [Hoskins and Hodges+05](https://doi.org/10.1175/JCLI3570.1); [Neu+13](https://doi.org/10.1175/BAMS-D-11-00154.1)].  The role of uncertainties due to the tracking method was highlighted by [[Neu+13](https://doi.org/10.1175/BAMS-D-11-00154.1)], who assessed 15 different tracking algorithms as part of an experiment set up by the international [Intercomparison of MId LAtitude STorm diagnostics (IMILAST)](https://proclim.scnat.ch/en/activities/project_imilast).

The following functions select data from a (Geo)DataFrame according to one or multiple conditions,
e.g. a specific year and tracking algorithm:

````python
def _select_from_column(df: pd.DataFrame, col: str, val) -> pd.Series:
    """
    Create a True/False mask for one column `col` in a (Geo)DataFrame `df`
    where the value is `val`.
    Checks for iterable/non-iterable values and switches between .isin and == accordingly.
    """
    # Check for single / multiple values
    if df[col].dtype == "O":  # String column
        if isinstance(val, str):
            # Value is str -> Single value
            iterable_val = False
        elif isinstance(val, Iterable):
            # Value is Iterable but not str -> assume Iterable[str]
            iterable_val = True
        else:
            raise ValueError(f"Cannot parse value of type `{type(val)}` for column `{col}`.")
    else:  # Non-string column, assume number or similar
        iterable_val = isinstance(val, Iterable)

    # Create True/False mask
    selection = (df[col].isin(val)) if iterable_val else (df[col] == val)
    return selection

def select_data(df: pd.DataFrame, **kwargs):
    """
    Select data from a given (Geo)DataFrame `df` according to any number of conditions,
    matching its column names.
    """
    # Create masks for all col/val pairs in kwargs, then combine with & (bitwise and)
    selection_all = [_select_from_column(df, col, val) for col, val in kwargs.items()]
    selection = reduce(bitwise_and, selection_all)

    # Apply and return
    df_selection = df[selection]
    return df_selection
````

The following functions allow large amounts of data to be downloaded from CDS, and to send requests based on track information.

````python
# Handling CDS size limits
def batch_requests(main_request: dict, *, batch_key: str="year", n: int=20) -> list[dict]:
    """ Take a big request (e.g. ERA5–Drought for all years) and separate it into smaller ones (size `n`). """
    from itertools import batched
    full_range = main_request[batch_key]  # e.g. [1940, 1941, ..., 2024]
    batched_range = batched(full_range, n)  # e.g. [1940, ..., 1959], [1960, ..., 1979], ...
    subrequests = [main_request | {batch_key: batch} for batch in batched_range]  # create corresponding CDS requests
    return subrequests
    
def build_era5msl_request_from_time(time_value):
    """
    Create a CDS API request for ERA5 MSLP for the given date.
    Always retrieves the 4 synoptic times.
    """
    
    # Ensure datetime
    time_value = pd.to_datetime(time_value)

    year = f"{time_value.year}"
    month = f"{time_value.month:02d}"
    day = f"{time_value.day:02d}"

    request = {
        "product_type": ["reanalysis"],
        "variable": ["mean_sea_level_pressure"],
        "year": [year],
        "month": [month],
        "day": [day],
        "time": ["00:00", "06:00", "12:00", "18:00"],
        "data_format": "netcdf",
        "download_format": "unarchived",
        "area": [70, -80, 30, 35],  # windstorm track domain
    }

    return request
````

##### Indicator calculations

````python
def compute_mean_by_nuts(data, value_column, nuts_levels=(0, 1, 2)):
    """
    Compute mean values grouped by (algorithm, threshold) for each NUTS level.
    """
    # Take mean across years (frequency).
    data_grouped = (
        data[[value_column, "NUTS_level"]]
        .groupby(level=("algorithm", "threshold", "region"))
        .mean(numeric_only=True)
    )

    # Split by NUTS level
    nuts_groups = {
        level: data_grouped.loc[data_grouped["NUTS_level"] == level]
        for level in nuts_levels
    }

    # Take mean across entire domain
    result = {
        level: df.groupby(level=("algorithm", "threshold")).mean(numeric_only=True)
        for level, df in nuts_groups.items()
    }

    return result
````

##### Track analysis

````python
def get_track_ends(df):
    """
    Generate a summary dataset of cyclone tracks by extracting start and end properties.
    """
    
    records = []  # will store one summary row per track

    # Loop over each algorithm (first level of MultiIndex)
    for algo, df_algo in df.groupby(level="algorithm"):

        # Within each algorithm, loop over individual tracks (by id)
        for track_id, group in df_algo.groupby(level="id"):

            # Ensure track points are in chronological order
            group = group.sort_values("time")

            # Extract first and last points of the track
            start = group.iloc[0]
            end   = group.iloc[-1]

            # Build a summary record for this track
            records.append({
                "algorithm": algo,          # which tracking algorithm
                "id": track_id,             # track identifier

                # start point info
                "start_time": start["time"],
                "start_lat":  start["latitude"],
                "start_lon":  start["longitude"],

                # end point info
                "end_time":   end["time"],
                "end_lat":    end["latitude"],
                "end_lon":    end["longitude"],

                # intensity metric across the track
                "mean_fg10": group["fg10"].mean(),  #  mean
                "min_mslp": group["msl"].min(),  #  min
            })

    # Return everything as a DataFrame (one row per track)
    return pd.DataFrame(records)

def get_monthly_count(tracks_start_end: pd.DataFrame):
    """
    Compute monthly track counts per algo (ordered by calendar month).
    """

    # Extract month name from start_time
    tracks_start_end["month"] = tracks_start_end["start_time"].dt.month_name()

    # COunt tracks per algo per month.
    monthly_counts = (
        tracks_start_end
        .groupby(["algorithm", "month"])
        .size()
        .unstack(fill_value=0)
    )

    # Ensure calendar order.
    month_order = [
        "January", "February", "March", "April", "May", "June",
        "July", "August", "September", "October", "November", "December"
    ]

    # Reorder by calendar order.
    monthly_counts = monthly_counts[month_order]

    return monthly_counts

def get_yearly_count(tracks_start_end: pd.DataFrame):
    """
    Compute yearly track count per algo.
    """

    # Extract year from start_time
    tracks_start_end["year"] = tracks_start_end["start_time"].dt.year

    # Count tracks per algo per year
    yearly_counts = (
        tracks_start_end
        .groupby(["algorithm", "year"])
        .size()
        .unstack(fill_value=0)
    )

    return yearly_counts

def track_stats(tracks_start_end: pd.DataFrame):
    """
    Compute summary stats of storm track durations by algo and season.

    Calculates duration (in hours) for each track then groups by algo and computes descriptive statistics.
    """

    # Work on copy.
    df = tracks_start_end.copy()

    # Define seasons
    seasons = {
        "ALL": None,
        "DJF": [12, 1, 2],
        "JJA": [6, 7, 8],
    }

    # Compute track duration by days
    df["duration_days"] = (
        df["end_time"] - df["start_time"]
    ).dt.total_seconds() / (24*60*60)

    # Stores results for each season.
    results = {}

    # Loop through each season.
    for season_name, months in seasons.items():

        # Select season based on seasons.
        if months is None:
            subset = df
        else:
            subset = df[df["start_time"].dt.month.isin(months)]

        # Group by tracking algo.
        grouped = subset.groupby("algorithm")

        # Compute summary stats for duration
        stats = grouped.agg(
            mean=("duration_days", "mean"),
            median=("duration_days", "median"),
            std=("duration_days", "std"),
            q25=("duration_days", lambda x: x.quantile(0.25)),
            q75=("duration_days", lambda x: x.quantile(0.75)),
        )
        # Store results for this season.
        results[season_name] = stats

    # Return dictionary of seasonal statistics.
    return results

def track_count_comparison(track_data: pd.DataFrame):
    """
    Compute total storm count by tracking algorithm

    Count unique storm IDs for each algo, returning styled dataframe.
    """

    # Reset index, group by algo and count unique storm IDS.
    total_track_counts = (
        track_data.reset_index()
        .groupby("algorithm")["id"]
        .nunique()
        .reset_index()
    )

    # Rename columns for clarity.
    total_track_counts.columns = ["Algorithm", "Total Storm Count"]

    # Define styling functino to colour-code for each algorithm.
    def color_algo(val):
        if val == "hodges":
            return "color: blue"
        elif val == "tempest_extremes":
            return "color: red"
        return ""

    # Apply styling to algo column only.
    return total_track_counts.style.map(color_algo, subset=["Algorithm"])

def bin_to_xarray(algo_grid: pd.DataFrame, lat_bins = np.arange(-5, 70, 2.5), lon_bins = np.arange(-80, 35, 2.5)):
 
    """
    Bin latitude/longitude track data onto a regular grid and return as an xarray DataArray.

    Groups points into specified lat/lon bins, sums track counts within each grid cell,
    and outputs a 2D DataArray with lat/lon coordinates.
    """
   
    algo_grid = algo_grid.copy()
    
    # Bin latitude and longitude into intervals
    algo_grid['lat_bin'] = pd.cut(algo_grid['latitude'], bins=lat_bins)
    algo_grid['lon_bin'] = pd.cut(algo_grid['longitude'], bins=lon_bins)

    # Aggregate track_count within each lat/lon bin
    algo_grid = algo_grid.groupby(["lat_bin", "lon_bin"])["track_count"].sum().reset_index()
    
    # Convert interval bins → numeric coordinates (midpoints)
    algo_grid['lat'] = algo_grid['lat_bin'].apply(lambda x: x.mid) # Find midpoint of coordinates
    algo_grid['lon'] = algo_grid['lon_bin'].apply(lambda x: x.mid)


    # Create 2D grid: rows = lat, columns = lon
    grid = algo_grid.groupby(['lat', 'lon'])['track_count'].sum().unstack()

    
    # Convert pandas grid → xarray DataArray
    algo_da = xr.DataArray(
        grid.values,
        coords={
            "lat": grid.index.values,
            "lon": grid.columns.values
        },
        dims=["lat", "lon"],
        name="track_count"
    )

    # Return binned, gridded data
    return algo_da

def compute_season_fractions(h_count: pd.DataFrame, te_count: pd.DataFrame):
    """
    Compute seasonal ratios and fractions between two algorithms.
    """

    # Define seasons
    winter_months = ["December", "January", "February"]
    summer_months = ["June", "July", "August"]

    # Hodges seasonal totals
    h_winter = h_count[winter_months].sum(axis=1).iloc[0]
    h_summer = h_count[summer_months].sum(axis=1).iloc[0]

    # Tempest seasonal totals
    te_winter = te_count[winter_months].sum(axis=1).iloc[0]
    te_summer = te_count[summer_months].sum(axis=1).iloc[0]
    
    # Cross-algorithm ratios
    winter = h_winter / te_winter
    summer = h_summer / te_summer
    
    # Within-algorithm seasonal fractions
    h_frac = h_winter / h_summer
    te_frac = te_winter / te_summer
    
    return winter, summer, h_frac, te_frac
````

Functions used to match tracks by overlapping timestamps and mean distance between overlapping points.

````python
def match_by_time(data: pd.DataFrame, TRACKING_ALGORITHMS=["hodges", "tempest_extremes"]):
    """
    Identify matching tracks based on temporal overlap.
    """
    
    # Split dataset by tracking algorithm
    df1, df2 = [data.loc[alg] for alg in TRACKING_ALGORITHMS]

    # Move 'id' from index to column and rename for clarity
    df1 = df1.reset_index(level="id").rename(columns={"id": "id_hodges"})
    df2 = df2.reset_index(level="id").rename(columns={"id": "id_tempest_extremes"})

    # Find overlapping timestamps between the two datasets
    time_overlap = pd.merge(
        df1, df2,
        on="time",
        how="inner",
        suffixes=[f"_{alg}" for alg in TRACKING_ALGORITHMS]
    )
    
    # Count overlapping time points for each track pair
    overlap_counts = (
        time_overlap
        .groupby(["id_hodges", "id_tempest_extremes"])["time"]
        .nunique()
    )
    
    results = []
    
    # Loop through each candidate pair
    for (hodges_id, tempest_id), overlap_points in overlap_counts.items():
    
        # Total points in each track
        hodges_points = data.xs(
            ("hodges", hodges_id), level=("algorithm", "id")
        )["time"].nunique()
    
        tempest_points = data.xs(
            ("tempest_extremes", tempest_id), level=("algorithm", "id")
        )["time"].nunique()
    
        # Average track length
        mean_points = (hodges_points + tempest_points) / 2
    
        # Fraction of overlapping points
        time_match = overlap_points / mean_points
    
        # Apply threshold for matching
        match_true = time_match >= 0.6
    
        # Store results
        results.append({
            "hodges_id": hodges_id,
            "tempest_id": tempest_id,
            "overlap_points": overlap_points,
            "hodges_points": hodges_points,
            "tempest_points": tempest_points,
            "time_match": time_match,
            "match": match_true
        })
    
    # Convert to DataFrame and filter matches
    results_df = pd.DataFrame(results)
    time_matches_only = results_df[results_df["match"]]

    return time_matches_only

def match_by_mean_distance(data, time_matches_only):
    """
    Refine matches using mean spatial distance between tracks.
    """

    # Work on a copy
    time_matches_only = time_matches_only.copy()

    # Split dataset by algorithm
    hodges = data.loc["hodges"]
    tempest = data.loc["tempest_extremes"]
    
    # Loop through matched track pairs
    for hodges_id, tempest_id in zip(
        time_matches_only["hodges_id"],
        time_matches_only["tempest_id"]
    ):
        distances = []

        # Extract individual tracks and align by time
        h_track = hodges.loc[hodges_id].set_index("time")
        t_track = tempest.loc[tempest_id].set_index("time")

        # Find common timestamps
        common_times = h_track.index.intersection(t_track.index)
    
        # Compute distance at each common timestep
        for t in common_times:
            h = h_track.loc[t]
            te = t_track.loc[t]

            distances.append(
                hs.haversine(
                    (h["latitude"], h["longitude"]),
                    (te["latitude"], te["longitude"]),
                    unit=hs.Unit.KILOMETERS
                )
            )

        # Assign mean distance to the pair
        if distances:
            time_matches_only.loc[
                (time_matches_only["hodges_id"] == hodges_id) &
                (time_matches_only["tempest_id"] == tempest_id),
                "distance"
            ] = np.mean(distances)

    # Sort by smallest distance (best match first)
    unique_matches = time_matches_only.sort_values("distance")

    # Ensure one-to-one matching (remove duplicates)
    unique_matches = unique_matches.drop_duplicates(subset="hodges_id", keep="first")
    unique_matches = unique_matches.drop_duplicates(subset="tempest_id", keep="first")
    
    return unique_matches
````
